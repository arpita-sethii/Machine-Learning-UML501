# Assignment 7 — SVM With Preprocessing
This notebook includes preprocessing (StandardScaler) to avoid perfect accuracy and improve evaluation.

```python
from sklearn import datasets
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix
from sklearn.preprocessing import StandardScaler

iris = datasets.load_iris()
X = iris.data
y = iris.target

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.20, random_state=42)

def evaluate_model(model, X_test, y_test):
    y_pred = model.predict(X_test)
    print("Accuracy:", accuracy_score(y_test, y_pred))
    print("Precision:", precision_score(y_test, y_pred, average='macro'))
    print("Recall:", recall_score(y_test, y_pred, average='macro'))
    print("F1-Score:", f1_score(y_test, y_pred, average='macro'))
    print("Confusion Matrix:\n", confusion_matrix(y_test, y_pred))

kernels = {
    "Linear": SVC(kernel="linear"),
    "Polynomial (degree=3)": SVC(kernel="poly", degree=3),
    "RBF": SVC(kernel="rbf")
}

for name, model in kernels.items():
    print(f"\n--- Kernel: {name} ---")
    model.fit(X_train, y_train)
    evaluate_model(model, X_test, y_test)
```

```python
bc = datasets.load_breast_cancer()
X_bc = bc.data
y_bc = bc.target

X_train_bc, X_test_bc, y_train_bc, y_test_bc = train_test_split(X_bc, y_bc, test_size=0.20, random_state=42)

svm_no_scaling = SVC(kernel="rbf")
svm_no_scaling.fit(X_train_bc, y_train_bc)
print("WITHOUT SCALING:")
print("Training Accuracy:", svm_no_scaling.score(X_train_bc, y_train_bc))
print("Testing Accuracy:", svm_no_scaling.score(X_test_bc, y_test_bc))

scaler2 = StandardScaler()
X_train_scaled = scaler2.fit_transform(X_train_bc)
X_test_scaled = scaler2.transform(X_test_bc)

svm_scaled = SVC(kernel="rbf")
svm_scaled.fit(X_train_scaled, y_train_bc)
print("\nWITH STANDARD SCALING:")
print("Training Accuracy:", svm_scaled.score(X_train_scaled, y_train_bc))
print("Testing Accuracy:", svm_scaled.score(X_test_scaled, y_test_bc))
```
